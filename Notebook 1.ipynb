```
import random

# Function to make a simple maths problem
# Example: "3 * x = 12"
def generate_equation():
    a = random.randint(1, 10)  # A random number for the question
    x = random.randint(1, 10)  # The answer (x)
    b = a * x  # Calculate the result
    return a, b, x

# Function to ask the user to solve the problem
def ask_question(a, b):
    print(f"What is x? {a} * x = {b}")  # Show the problem
    try:
        user_answer = int(input("Type your answer: "))  # Get user input
        return user_answer
    except ValueError:  # If the input is not a number
        print("That's not a number. Try again next time!")
        return None

# Main part of the game
if __name__ == "__main__":
    print("Hi! Let's play a maths game!")  # Greeting
    num_questions = 5  # Number of problems
    score = 0  # Start score at zero

    for i in range(num_questions):
        print(f"\nProblem {i + 1}:")  # Tell the user the problem number
        a, b, correct_x = generate_equation()  # Create a problem
        user_answer = ask_question(a, b)  # Ask the question

        if user_answer is None:  # Skip if input was invalid
            print("Moving to the next problem.")
            continue

        if user_answer == correct_x:  # Check the answer
            print("Correct! Good job!")
            score += 1  # Add a point
        else:
            print(f"Oops! The answer was {correct_x}.")

    # Show the final score
    print(f"Game over! You got {score} out of {num_questions} right. Well done!")
```
